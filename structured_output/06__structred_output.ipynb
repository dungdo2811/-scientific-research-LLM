```python
import json
from pathlib import Path

query = {
    "locations": ["Hà Nội"],
    "indicator": "PM2.5",
    "operator": "gt",
    "value": 50.0,
    "include_inactive": False,
}

json_text = json.dumps(
    query,
    ensure_ascii= False,
    indent= 2,
)
print(json_text)
```

```python

print(json_text)
Path("query.json").write_text(json_text,encoding = "utf-8")
loaded= json.loads(
Path("query.json").read_text(encoding = "utf-8")
)

print(loaded)
```
